```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.read_csv("house_price_regression_dataset.csv")

X = df[["Square_Footage"]]
y = df["Year_Built"]

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

r2 = r2_score(y, y_pred)

print("R2 Score (Square_Footage -> Year_Built):", r2)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.read_csv("house_price_regression_dataset.csv")

# Model 1: age -> membership_years
X1 = df[["Year_Built"]]
y1 = df["Square_Footage"]

model1 = LinearRegression()
model1.fit(X1, y1)

r2_1 = r2_score(y1, model1.predict(X1))


# Model 2: age -> number_of_children
X2 = df[["Num_Bedrooms"]]
y2 = df["Lot_Size"]

model2 = LinearRegression()
model2.fit(X2, y2)

r2_2 = r2_score(y2, model2.predict(X2))


# Model 3: membership_years -> number_of_children
X3 = df[["Garage_Size"]]
y3 = df["Neighborhood_Quality"]

model3 = LinearRegression()
model3.fit(X3, y3)

r2_3 = r2_score(y3, model3.predict(X3))


# Display results
results = pd.DataFrame({
    "Regression Model": [
        "Square_Footage -> Year_Built",
        "Num_Bedrooms -> Lot_Size",
        "Garage_Size -> Neighborhood_Quality"
    ],
    "R2 Score": [
        r2_1,
        r2_2,
        r2_3
    ]
})

print(results)

# Find best model
best_model = results.loc[
    results["R2 Score"].idxmax()
]

print("\nBest Regression Model:")
print(best_model)
```
