```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import ttest_1samp
from statsmodels.stats.proportion import proportions_ztest


# PART A: HYPOTHESIS TESTING

# Q1. Load the supermarket sales dataset
df = pd.read_csv("supermarket_sales.csv")

# Q2. Display first five records
print("\nQ2. First Five Records:")
print(df.head())

# Q3. Select revenue column
revenue = df["Total"]

print("\nQ3. Revenue Column:")
print(revenue)

# Q4. Check for missing values
print("\nQ4. Missing Values:")
print(revenue.isnull().sum())

# Q5. Calculate mean revenue
mean_revenue = revenue.mean()
print("\nQ5. Mean Revenue:")
print(mean_revenue)

# Q6. Expected average revenue
expected_revenue = 1500
print("\nQ6. Expected Average Revenue:")
print(expected_revenue)

# Q7. Null Hypothesis
print("\nQ7. Null Hypothesis:")
print("H0: μ = 1500")

# Q8. Alternative Hypothesis
print("\nQ8. Alternative Hypothesis:")
print("H1: μ ≠ 1500")

# Q9. Significance level
alpha = 0.05
print("\nQ9. Significance Level:")
print(alpha)

# Q10. One-sample t-test
t_stat, p_value = ttest_1samp(revenue, expected_revenue)

print("\nQ10. One-Sample T-Test:")
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Q11. P-value
print("\nQ11. P-value:")
print(p_value)

# Q12. Compare p-value with 0.05
print("\nQ12. Comparison:")
if p_value < alpha:
    print("P-value < 0.05")
else:
    print("P-value >= 0.05")

# Q13. Decision
print("\nQ13. Decision:")
if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Do Not Reject the Null Hypothesis")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.stats.proportion import proportions_ztest

# Q1. Load the A/B testing dataset
df = pd.read_csv("ab_testing.csv")

# Q2. Display the data
print("A/B Testing Data:")
print(df)

# Q3. Calculate conversion rate for Group A
group_a = df[df["Group"] == "A"].iloc[0]
rate_a = group_a["Conversions"] / group_a["Users"]

print("\nQ3. Conversion Rate for Group A:")
print(rate_a * 100, "%")

# Q4. Calculate conversion rate for Group B
group_b = df[df["Group"] == "B"].iloc[0]
rate_b = group_b["Conversions"] / group_b["Users"]

print("\nQ4. Conversion Rate for Group B:")
print(rate_b * 100, "%")

# Q5. Compare conversion rates
print("\nQ5. Comparison:")
print("Group A:", rate_a * 100, "%")
print("Group B:", rate_b * 100, "%")

# Q6. H0: pA = pB

# Q7. H1: pA ≠ pB

# Q8. Significance level
alpha = 0.05

# Q9. Perform two-proportion test
conversions = [group_a["Conversions"], group_b["Conversions"]]
users = [group_a["Users"], group_b["Users"]]

z_stat, p_value = proportions_ztest(conversions, users)

print("\nQ9. Two-Proportion Test:")
print("Z-statistic:", z_stat)

# Q10. Calculate p-value
print("\nQ10. P-value:")
print(p_value)

# Q11. Compare p-value with 0.05
print("\nQ11. Comparison:")
if p_value < 0.05:
    print("P-value < 0.05")
else:
    print("P-value >= 0.05")

# Q12. Decision
print("\nQ12. Decision:")
if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Do Not Reject the Null Hypothesis")

# Q13. Create bar chart
groups = ["A", "B"]
conversion_rates = [rate_a * 100, rate_b * 100]

plt.bar(groups, conversion_rates)
plt.xlabel("Group")
plt.ylabel("Conversion Rate (%)")
plt.title("A/B Testing Conversion Rates")
plt.show()
```
