```python
# Porter Business Analysis Project

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.optimize import linprog
from statsmodels.stats.proportion import proportions_ztest
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Load dataset

df = pd.read_csv("Porter_Tableau_Dashboard_Dataset.csv")

# Display basic information

print("First 5 Records")
print(df.head())

print("\nDataset Shape")
print(df.shape)

print("\nColumn Names")
print(df.columns.tolist())

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Records")
print(df.duplicated().sum())

print("\nStatistical Summary")
print(df.describe(include="all"))

# Data preprocessing

df = df.drop_duplicates()

numeric_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(include=["object", "category"]).columns

print("\nNumerical Columns")
print(list(numeric_cols))

print("\nCategorical Columns")
print(list(categorical_cols))

# Missing value handling

for column in numeric_cols:
    df[column] = df[column].fillna(df[column].median())

for column in categorical_cols:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

print("\nMissing Values After Cleaning")
print(df.isnull().sum())

# Numerical analysis

print("\nMean")
print(df[numeric_cols].mean())

print("\nMedian")
print(df[numeric_cols].median())

print("\nStandard Deviation")
print(df[numeric_cols].std())

print("\nMinimum Values")
print(df[numeric_cols].min())

print("\nMaximum Values")
print(df[numeric_cols].max())

# Categorical analysis

for column in categorical_cols:

    print("\nValue Counts for", column)
    print(df[column].value_counts().head(10))

# Numerical distributions

for column in numeric_cols:

    plt.figure(figsize=(8, 5))
    plt.hist(df[column], bins=20)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title("Distribution of " + column)
    plt.show()

# Boxplots for numerical columns

for column in numeric_cols:

    plt.figure(figsize=(8, 5))
    plt.boxplot(df[column])
    plt.ylabel(column)
    plt.title("Boxplot of " + column)
    plt.show()

# Categorical visualizations

for column in categorical_cols:

    if df[column].nunique() <= 15:

        plt.figure(figsize=(10, 5))
        df[column].value_counts().plot(kind="bar")
        plt.xlabel(column)
        plt.ylabel("Count")
        plt.title("Distribution of " + column)
        plt.xticks(rotation=45)
        plt.show()

# Correlation analysis

if len(numeric_cols) >= 2:

    correlation = df[numeric_cols].corr()

    print("\nCorrelation Matrix")
    print(correlation)

    plt.figure(figsize=(12, 7))
    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )
    plt.title("Correlation Matrix")
    plt.show()

# Latitude and Longitude analysis

latitude_col = None
longitude_col = None

for column in df.columns:

    if column.lower() in ["latitude", "lat"]:
        latitude_col = column

    if column.lower() in ["longitude", "long", "lng"]:
        longitude_col = column

if latitude_col is not None and longitude_col is not None:

    print("\nLatitude Summary")
    print(df[latitude_col].describe())

    print("\nLongitude Summary")
    print(df[longitude_col].describe())

    plt.figure(figsize=(9, 7))

    plt.scatter(
        df[longitude_col],
        df[latitude_col],
        alpha=0.5
    )

    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.title("Porter Delivery Locations")
    plt.show()

# Scatter plots between numerical variables

if len(numeric_cols) >= 2:

    x_column = numeric_cols[0]
    y_column = numeric_cols[1]

    plt.figure(figsize=(8, 5))

    plt.scatter(
        df[x_column],
        df[y_column],
        alpha=0.5
    )

    plt.xlabel(x_column)
    plt.ylabel(y_column)
    plt.title(x_column + " vs " + y_column)
    plt.show()

# Hypothesis testing

print("\nHypothesis Testing")

if len(numeric_cols) > 0:

    test_column = numeric_cols[0]

    data = df[test_column].dropna()

    expected_mean = data.median()

    t_stat, p_value = stats.ttest_1samp(
        data,
        expected_mean
    )

    print("Testing Column:", test_column)
    print("Expected Mean:", expected_mean)
    print("Sample Mean:", data.mean())
    print("T-Statistic:", t_stat)
    print("P-Value:", p_value)

    alpha = 0.05

    if p_value < alpha:
        print("Result: Reject the Null Hypothesis")
    else:
        print("Result: Fail to Reject the Null Hypothesis")

# A/B Testing

# Proposed Porter business experiment

print("\nA/B Testing")

group_a_orders = 1000
group_a_success = 820

group_b_orders = 1000
group_b_success = 870

group_a_rate = group_a_success / group_a_orders
group_b_rate = group_b_success / group_b_orders

print("Group A Success Rate:", group_a_rate)
print("Group B Success Rate:", group_b_rate)

successes = np.array([
    group_a_success,
    group_b_success
])

trials = np.array([
    group_a_orders,
    group_b_orders
])

z_stat, p_value = proportions_ztest(
    successes,
    trials
)

print("Z-Statistic:", z_stat)
print("P-Value:", p_value)

if p_value < 0.05:
    print("Result: There is a significant difference between Group A and Group B")
else:
    print("Result: There is no significant difference between Group A and Group B")

plt.figure(figsize=(7, 5))

plt.bar(
    ["Group A", "Group B"],
    [group_a_rate, group_b_rate]
)

plt.xlabel("Group")
plt.ylabel("Success Rate")
plt.title("Porter A/B Testing")
plt.show()

# Optimization

# Proposed Porter vehicle allocation scenario

print("\nOptimization")

print("Area A Profit per Vehicle = Rs. 500")
print("Area B Profit per Vehicle = Rs. 700")

profit = [-500, -700]

constraints = [
    [1, 1],
    [1, 0],
    [0, 1]
]

limits = [
    100,
    60,
    70
]

optimization = linprog(
    profit,
    A_ub=constraints,
    b_ub=limits,
    bounds=[
        (0, None),
        (0, None)
    ],
    method="highs"
)

if optimization.success:

    area_a_vehicles = optimization.x[0]
    area_b_vehicles = optimization.x[1]

    maximum_profit = -optimization.fun

    print("Vehicles for Area A:", area_a_vehicles)
    print("Vehicles for Area B:", area_b_vehicles)
    print("Maximum Daily Profit: Rs.", maximum_profit)

# Simulation

# Proposed Porter order simulation

print("\nSimulation")

np.random.seed(42)

number_of_days = 1000

simulated_orders = np.random.normal(
    loc=500,
    scale=80,
    size=number_of_days
)

simulated_orders = np.maximum(
    simulated_orders,
    0
)

average_orders = simulated_orders.mean()
average_revenue = average_orders * 120

print("Number of Simulated Days:", number_of_days)
print("Average Orders Per Day:", average_orders)
print("Estimated Revenue Per Day: Rs.", average_revenue)

plt.figure(figsize=(9, 5))

plt.hist(
    simulated_orders,
    bins=30
)

plt.xlabel("Orders Per Day")
plt.ylabel("Frequency")
plt.title("Porter Daily Order Simulation")
plt.show()

# Experimental and factorial design

# Proposed Porter business experiment

print("\nFactorial Design")

print("Factor A = Delivery Fee")
print("Factor B = Discount")
print("Response = Number of Orders")

factorial_data = pd.DataFrame({
    "Fee": [
        "Low",
        "Low",
        "High",
        "High",
        "Low",
        "Low",
        "High",
        "High"
    ],
    "Discount": [
        "Low",
        "High",
        "Low",
        "High",
        "Low",
        "High",
        "Low",
        "High"
    ],
    "Orders": [
        420,
        510,
        350,
        450,
        430,
        520,
        360,
        455
    ]
})

print("\nFactorial Experiment Data")
print(factorial_data)

model = ols(
    "Orders ~ C(Fee) + C(Discount) + C(Fee):C(Discount)",
    data=factorial_data
).fit()

anova_result = sm.stats.anova_lm(
    model,
    typ=2
)

print("\nANOVA Result")
print(anova_result)

plt.figure(figsize=(8, 5))

for discount in factorial_data["Discount"].unique():

    subset = factorial_data[
        factorial_data["Discount"] == discount
    ]

    average_orders_by_fee = subset.groupby(
        "Fee"
    )["Orders"].mean()

    plt.plot(
        average_orders_by_fee.index,
        average_orders_by_fee.values,
        marker="o",
        label="Discount " + discount
    )

plt.xlabel("Delivery Fee")
plt.ylabel("Average Orders")
plt.title("Porter Factorial Design")
plt.legend()
plt.show()

# Final results

print("\nFinal Project Results")

print("The Porter dataset was analyzed using descriptive statistics,")
print("data preprocessing, visualization, correlation analysis,")
print("hypothesis testing and location analysis.")

print("\nA/B testing was performed as a proposed Porter business")
print("experiment to compare two delivery processes.")

print("\nOptimization was performed as a proposed business scenario")
print("to find the best vehicle allocation for maximum profit.")

print("\nSimulation was performed to estimate daily Porter orders")
print("and expected revenue under uncertain demand.")

print("\nFactorial design was performed as a proposed experiment")
print("to study the effect of delivery fee and discount on orders.")

print("\nThe original Porter CSV dataset was not modified.")
```
