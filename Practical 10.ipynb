```python


import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error


# ------------------------------------------------------------
# Q1. Load the sales_data.csv dataset
# ------------------------------------------------------------

df = pd.read_csv("sales_data.csv")

print("Dataset loaded successfully!")


# ------------------------------------------------------------
# Q2. Display the first five records
# ------------------------------------------------------------

print("\nFirst Five Records:")
print(df.head())


# ------------------------------------------------------------
# Q3. Identify independent and dependent variables
# ------------------------------------------------------------

print("\nVariables:")
print("Independent Variable: advertising")
print("Dependent Variable: sales")


# ------------------------------------------------------------
# Q4. Select advertising as independent variable
#     and sales as dependent variable
# ------------------------------------------------------------

X = df[["advertising"]]
y = df["sales"]

print("\nIndependent Variable:")
print(X.head())

print("\nDependent Variable:")
print(y.head())


# ------------------------------------------------------------
# Q5. Check data for missing values
# ------------------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

if df.isnull().sum().sum() == 0:
    print("No missing values found.")
else:
    print("Missing values are present.")


# ------------------------------------------------------------
# Q6. Create scatter plot
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.scatter(
    df["advertising"],
    df["sales"]
)

plt.xlabel("Advertising Expenditure")
plt.ylabel("Sales")
plt.title("Advertising Expenditure vs Sales")
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# Q7. Develop a simple linear regression model
# ------------------------------------------------------------

model = LinearRegression()

model.fit(X, y)

print("\nLinear Regression Model created successfully.")


# ------------------------------------------------------------
# Q8. Display the regression equation
# ------------------------------------------------------------

b0 = model.intercept_
b1 = model.coef_[0]

print("\nRegression Equation:")
print(f"Sales = {b0:.2f} + ({b1:.2f} × Advertising)")


# ------------------------------------------------------------
# Q9. Predict sales for a given advertising expenditure
# ------------------------------------------------------------

advertising_value = 60

predicted_sales = model.predict([[advertising_value]])

print("\nPrediction:")
print("Advertising Expenditure =", advertising_value)
print("Predicted Sales =", round(predicted_sales[0], 2))


# ------------------------------------------------------------
# Q10. Calculate R2 score
# ------------------------------------------------------------

y_pred = model.predict(X)

r2 = r2_score(y, y_pred)

print("\nR2 Score:")
print(round(r2, 4))


# ------------------------------------------------------------
# Q11. Calculate Mean Squared Error
# ------------------------------------------------------------

mse = mean_squared_error(y, y_pred)

print("\nMean Squared Error:")
print(round(mse, 4))


# ------------------------------------------------------------
# Q12. Compare actual sales with predicted sales
# ------------------------------------------------------------

comparison = pd.DataFrame({
    "Advertising": df["advertising"],
    "Actual Sales": y,
    "Predicted Sales": y_pred.round(2)
})

print("\nActual vs Predicted Sales:")
print(comparison)


# ------------------------------------------------------------
# Q13. Plot regression line with actual data points
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

# Actual data points
plt.scatter(
    df["advertising"],
    df["sales"],
    label="Actual Data"
)

# Regression line
plt.plot(
    df["advertising"],
    y_pred,
    label="Regression Line"
)

plt.xlabel("Advertising Expenditure")
plt.ylabel("Sales")
plt.title("Linear Regression: Advertising vs Sales")

plt.legend()
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# Q14. Interpret the regression results
# ------------------------------------------------------------

print("\nRegression Interpretation:")
print("--------------------------")

print("Intercept (b0):", round(b0, 2))
print("Slope (b1):", round(b1, 2))
print("R2 Score:", round(r2, 4))
print("MSE:", round(mse, 4))

if b1 > 0:
    print("\nThere is a positive relationship between")
    print("advertising expenditure and sales.")
    print("As advertising expenditure increases,")
    print("sales tend to increase.")
else:
    print("\nThere is a negative relationship between")
    print("advertising expenditure and sales.")


# ------------------------------------------------------------
# Q15. Conclusion
# ------------------------------------------------------------

print("\nConclusion:")
print("The regression analysis shows that advertising")
print("expenditure has a positive relationship with sales.")
print("Higher advertising expenditure is associated")
print("with higher sales in the given dataset.")

print("\nPART A COMPLETED SUCCESSFULLY!")
```

```python
# ============================================================
# PRACTICAL 10 - PART B
# TIME SERIES ANALYSIS
# Dataset: monthly_sales.csv
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import mean_squared_error

from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.arima.model import ARIMA


# ------------------------------------------------------------
# Q16. Load monthly sales data
# ------------------------------------------------------------

monthly_df = pd.read_csv("monthly_sales.csv")

print("Monthly Sales Data:")
print(monthly_df)


# ------------------------------------------------------------
# Q17. Convert date column into date format
# ------------------------------------------------------------

monthly_df["Month"] = pd.to_datetime(monthly_df["Month"])

# Set Month as index
monthly_df.set_index("Month", inplace=True)

print("\nDate Column Converted:")
print(monthly_df)

print("\nData Types:")
print(monthly_df.dtypes)


# ------------------------------------------------------------
# Q18. Plot monthly sales using a line chart
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_df.index,
    monthly_df["Sales"],
    marker="o"
)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")

plt.grid(True)
plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# Q19. Observe the trend
# ------------------------------------------------------------

first_sales = monthly_df["Sales"].iloc[0]
last_sales = monthly_df["Sales"].iloc[-1]

print("\nTrend Analysis:")

if last_sales > first_sales:
    print("The overall sales trend is increasing.")
elif last_sales < first_sales:
    print("The overall sales trend is decreasing.")
else:
    print("The sales trend is stable.")

print("First Month Sales:", first_sales)
print("Last Month Sales:", last_sales)


# ------------------------------------------------------------
# Q20. Apply Exponential Smoothing
# ------------------------------------------------------------

sales_series = monthly_df["Sales"]

exp_model = ExponentialSmoothing(
    sales_series,
    trend="add",
    seasonal=None
)

exp_fit = exp_model.fit()

print("\nExponential Smoothing applied successfully.")

print("\nFitted Values:")
print(exp_fit.fittedvalues)


# ------------------------------------------------------------
# Q21. Forecast the next 3 months
# ------------------------------------------------------------

exp_forecast = exp_fit.forecast(3)

# Create next 3 month dates
future_dates = pd.date_range(
    start=monthly_df.index[-1] + pd.DateOffset(months=1),
    periods=3,
    freq="MS"
)

exp_forecast_df = pd.DataFrame({
    "Month": future_dates,
    "Forecast": exp_forecast.values
})

print("\nExponential Smoothing Forecast:")
print(exp_forecast_df)


# ------------------------------------------------------------
# Q22. Plot actual and forecasted sales
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

# Actual sales
plt.plot(
    monthly_df.index,
    monthly_df["Sales"],
    marker="o",
    label="Actual Sales"
)

# Forecast
plt.plot(
    future_dates,
    exp_forecast.values,
    marker="o",
    linestyle="--",
    label="Exponential Smoothing Forecast"
)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Actual and Forecasted Sales")

plt.legend()
plt.grid(True)
plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# Q23. Calculate forecasting error
# ------------------------------------------------------------

exp_fitted = exp_fit.fittedvalues

exp_mse = mean_squared_error(
    sales_series,
    exp_fitted
)

exp_rmse = np.sqrt(exp_mse)

print("\nExponential Smoothing Forecast Error:")
print("MSE =", round(exp_mse, 4))
print("RMSE =", round(exp_rmse, 4))


# ------------------------------------------------------------
# Q24. Apply ARIMA to the same data
# ------------------------------------------------------------

arima_model = ARIMA(
    sales_series,
    order=(1, 1, 1)
)

arima_fit = arima_model.fit()

print("\nARIMA Model Applied Successfully.")

print("\nARIMA Model Summary:")
print(arima_fit.summary())


# ------------------------------------------------------------
# Q25. Forecast next 3 months using ARIMA
# ------------------------------------------------------------

arima_forecast = arima_fit.forecast(
    steps=3
)

arima_forecast_df = pd.DataFrame({
    "Month": future_dates,
    "Forecast": arima_forecast.values
})

print("\nARIMA Forecast:")
print(arima_forecast_df)


# ------------------------------------------------------------
# Q26. Compare ARIMA and Exponential Smoothing forecasts
# ------------------------------------------------------------

forecast_comparison = pd.DataFrame({
    "Month": future_dates,
    "Exponential Smoothing": exp_forecast.values,
    "ARIMA": arima_forecast.values
})

print("\nForecast Comparison:")
print(forecast_comparison)


# ------------------------------------------------------------
# Q27. Calculate and compare forecasting errors
# ------------------------------------------------------------

# Exponential Smoothing error
exp_fitted = exp_fit.fittedvalues

exp_mse = mean_squared_error(
    sales_series,
    exp_fitted
)

exp_rmse = np.sqrt(exp_mse)


# ARIMA fitted values
arima_fitted = arima_fit.fittedvalues

# Remove missing values
valid_data = pd.concat(
    [
        sales_series.rename("Actual"),
        arima_fitted.rename("Predicted")
    ],
    axis=1
).dropna()

arima_mse = mean_squared_error(
    valid_data["Actual"],
    valid_data["Predicted"]
)

arima_rmse = np.sqrt(arima_mse)


# Compare errors
error_comparison = pd.DataFrame({
    "Model": [
        "Exponential Smoothing",
        "ARIMA"
    ],
    "MSE": [
        exp_mse,
        arima_mse
    ],
    "RMSE": [
        exp_rmse,
        arima_rmse
    ]
})

print("\nForecasting Error Comparison:")
print(error_comparison)


# ------------------------------------------------------------
# Q28. Identify model with lower forecasting error
# ------------------------------------------------------------

print("\nModel with Lower Forecasting Error:")

if exp_rmse < arima_rmse:
    print("Exponential Smoothing has the lower RMSE.")
elif arima_rmse < exp_rmse:
    print("ARIMA has the lower RMSE.")
else:
    print("Both models have the same RMSE.")


# ------------------------------------------------------------
# Q29. Final conclusion
# ------------------------------------------------------------

print("\n==========================================")
print("FINAL CONCLUSION")
print("==========================================")

print("1. Monthly sales show an overall increasing trend.")

print("2. Exponential Smoothing was applied")
print("   to forecast the next 3 months.")

print("3. ARIMA was applied to forecast")
print("   the next 3 months.")

print("4. Both forecasting models were compared")
print("   using MSE and RMSE.")

if exp_rmse < arima_rmse:
    print("5. Exponential Smoothing produced the")
    print("   lower forecasting error for this dataset.")
elif arima_rmse < exp_rmse:
    print("5. ARIMA produced the lower forecasting")
    print("   error for this dataset.")
else:
    print("5. Both models produced the same error.")

print("\nPART B COMPLETED SUCCESSFULLY!")
```
