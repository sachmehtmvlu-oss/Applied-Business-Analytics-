```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================
# A) DICE ROLL PROBABILITY EXPERIMENT
# ============================================

np.random.seed(42)

# 1. Simulate 1,000 dice rolls
rolls = np.random.randint(1, 7, 1000)

print("=" * 60)
print("A) DICE ROLL PROBABILITY EXPERIMENT")
print("=" * 60)

print("\n1. First 20 dice outcomes:")
print(rolls[:20])

# 2. Experimental probability
frequency = np.bincount(rolls, minlength=7)[1:]
experimental_probability = frequency / 1000

print("\n2. Experimental Probability:")
for i in range(6):
    print(f"Number {i+1}: Frequency = {frequency[i]}, "
          f"Probability = {experimental_probability[i]:.4f}")

# 3. Theoretical probability
theoretical_probability = 1 / 6

print("\n3. Theoretical Probability:")
for i in range(1, 7):
    print(f"Number {i}: {theoretical_probability:.4f}")

# 4. Probability of getting 1,2,3,4,5,6
print("\n4. Probability of getting each number:")
for i in range(6):
    print(f"P({i+1}) = {experimental_probability[i]:.4f}")

# 5. Even number
even_probability = np.mean(rolls % 2 == 0)
print(f"\n5. Probability of getting an even number: {even_probability:.4f}")

# 6. Odd number
odd_probability = np.mean(rolls % 2 != 0)
print(f"6. Probability of getting an odd number: {odd_probability:.4f}")

# 7. Greater than 4
greater_4_probability = np.mean(rolls > 4)
print(f"7. Probability of getting a number greater than 4: "
      f"{greater_4_probability:.4f}")

# 8. Less than 4
less_4_probability = np.mean(rolls < 4)
print(f"8. Probability of getting a number less than 4: "
      f"{less_4_probability:.4f}")

# 9. Greater than or equal to 4
greater_equal_4_probability = np.mean(rolls >= 4)
print(f"9. Probability of getting a number >= 4: "
      f"{greater_equal_4_probability:.4f}")

# 10. Frequency bar chart
plt.figure(figsize=(8, 5))
plt.bar(range(1, 7), frequency)
plt.xlabel("Dice Outcome")
plt.ylabel("Frequency")
plt.title("Frequency of Dice Outcomes")
plt.xticks(range(1, 7))
plt.show()

# 11. Experimental probability bar chart
plt.figure(figsize=(8, 5))
plt.bar(range(1, 7), experimental_probability)
plt.xlabel("Dice Outcome")
plt.ylabel("Experimental Probability")
plt.title("Experimental Probability of Dice Outcomes")
plt.xticks(range(1, 7))
plt.show()

# 12. Pie chart
plt.figure(figsize=(7, 7))
plt.pie(
    frequency,
    labels=[1, 2, 3, 4, 5, 6],
    autopct="%1.1f%%"
)
plt.title("Percentage Distribution of Dice Outcomes")
plt.show()

# 13. Experimental vs theoretical probability
plt.figure(figsize=(8, 5))
plt.plot(
    range(1, 7),
    experimental_probability,
    marker="o",
    label="Experimental Probability"
)
plt.plot(
    range(1, 7),
    [theoretical_probability] * 6,
    marker="o",
    label="Theoretical Probability"
)
plt.xlabel("Dice Outcome")
plt.ylabel("Probability")
plt.title("Experimental vs Theoretical Probability")
plt.xticks(range(1, 7))
plt.legend()
plt.grid(True)
plt.show()

# 14. Even vs Odd frequency
even_frequency = np.sum(rolls % 2 == 0)
odd_frequency = np.sum(rolls % 2 != 0)

plt.figure(figsize=(7, 5))
plt.bar(["Even", "Odd"], [even_frequency, odd_frequency])
plt.xlabel("Type")
plt.ylabel("Frequency")
plt.title("Frequency of Even vs Odd Outcomes")
plt.show()

# 15. Experimental probabilities
categories = [
    "Even",
    "Odd",
    "> 4",
    "< 4"
]

probabilities = [
    even_probability,
    odd_probability,
    greater_4_probability,
    less_4_probability
]

plt.figure(figsize=(8, 5))
plt.bar(categories, probabilities)
plt.xlabel("Outcome Category")
plt.ylabel("Experimental Probability")
plt.title("Experimental Probability of Different Outcomes")
plt.show()
```

```python
# ============================================
# B) PROBABILITY DISTRIBUTIONS
# ============================================

import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

np.random.seed(42)

print("\n" + "=" * 60)
print("B) PROBABILITY DISTRIBUTIONS")
print("=" * 60)

# --------------------------------------------
# 1. Normal Distribution
# --------------------------------------------

normal_data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)

print("\n1. Normal Distribution")
print("Mean = 50")
print("Standard Deviation = 10")

# 2. Mean, Median, Mode, Variance, SD
mean_normal = np.mean(normal_data)
median_normal = np.median(normal_data)

# Approximate mode using histogram
hist, bins = np.histogram(normal_data, bins=30)
mode_bin = np.argmax(hist)
mode_normal = (bins[mode_bin] + bins[mode_bin + 1]) / 2

variance_normal = np.var(normal_data)
std_normal = np.std(normal_data)

print("\n2. Mean, Median, Mode, Variance and Standard Deviation")
print(f"Mean       : {mean_normal:.4f}")
print(f"Median     : {median_normal:.4f}")
print(f"Mode       : {mode_normal:.4f}")
print(f"Variance   : {variance_normal:.4f}")
print(f"Std Dev    : {std_normal:.4f}")

# 3. Complete statistics
print("\n3. Statistics of Normal Distribution")
print(f"Mean       : {np.mean(normal_data):.4f}")
print(f"Median     : {np.median(normal_data):.4f}")
print(f"Variance   : {np.var(normal_data):.4f}")
print(f"Std Dev    : {np.std(normal_data):.4f}")
print(f"Minimum    : {np.min(normal_data):.4f}")
print(f"Maximum    : {np.max(normal_data):.4f}")

# 4. Histogram
plt.figure(figsize=(8, 5))
plt.hist(normal_data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normal Distribution")
plt.show()

print("\n4. Histogram Analysis:")
print("The histogram is approximately bell-shaped and symmetric around 50.")

# 5. Box plot
plt.figure(figsize=(8, 5))
plt.boxplot(normal_data)
plt.ylabel("Value")
plt.title("Box Plot of Normal Distribution")
plt.show()

# 6. Percentage between 40 and 60
between_40_60 = np.mean(
    (normal_data >= 40) & (normal_data <= 60)
) * 100

print(f"\n6. Percentage between 40 and 60: {between_40_60:.2f}%")

# 7. Percentage between 30 and 70
between_30_70 = np.mean(
    (normal_data >= 30) & (normal_data <= 70)
) * 100

print(f"7. Percentage between 30 and 70: {between_30_70:.2f}%")

# --------------------------------------------
# Uniform Distribution
# --------------------------------------------

# 8. Generate uniform distribution
uniform_data = np.random.uniform(
    0,
    100,
    1000
)

print("\n8. Uniform Distribution")
print("Generated 1,000 values between 0 and 100.")

print(f"Mean   : {np.mean(uniform_data):.4f}")
print(f"Median : {np.median(uniform_data):.4f}")
print(f"Min    : {np.min(uniform_data):.4f}")
print(f"Max    : {np.max(uniform_data):.4f}")

# 9. Uniform histogram
plt.figure(figsize=(8, 5))
plt.hist(uniform_data, bins=20)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Uniform Distribution")
plt.show()

# 10. Uniform box plot
plt.figure(figsize=(8, 5))
plt.boxplot(uniform_data)
plt.ylabel("Value")
plt.title("Box Plot of Uniform Distribution")
plt.show()

# --------------------------------------------
# Exponential Distribution
# --------------------------------------------

# 11. Generate exponential distribution
exponential_data = np.random.exponential(
    scale=2,
    size=1000
)

print("\n11. Exponential Distribution")
print("Scale = 2")

print(f"Mean       : {np.mean(exponential_data):.4f}")
print(f"Median     : {np.median(exponential_data):.4f}")
print(f"Variance   : {np.var(exponential_data):.4f}")
print(f"Std Dev    : {np.std(exponential_data):.4f}")
print(f"Minimum    : {np.min(exponential_data):.4f}")
print(f"Maximum    : {np.max(exponential_data):.4f}")

# 12. Exponential histogram
plt.figure(figsize=(8, 5))
plt.hist(exponential_data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Exponential Distribution")
plt.show()

# 13. Exponential box plot
plt.figure(figsize=(8, 5))
plt.boxplot(exponential_data)
plt.ylabel("Value")
plt.title("Box Plot of Exponential Distribution")
plt.show()

# Identify possible extreme values using IQR
Q1 = np.percentile(exponential_data, 25)
Q3 = np.percentile(exponential_data, 75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

extreme_values = exponential_data[
    (exponential_data < lower_limit) |
    (exponential_data > upper_limit)
]

print("\n13. Possible Extreme Values")
print(f"Q1              : {Q1:.4f}")
print(f"Q3              : {Q3:.4f}")
print(f"IQR             : {IQR:.4f}")
print(f"Upper Limit     : {upper_limit:.4f}")
print(f"Extreme Values  : {len(extreme_values)}")

# 14. Percentage greater than 2
greater_2 = np.mean(exponential_data > 2) * 100

print(f"\n14. Percentage greater than 2: {greater_2:.2f}%")

# 15. Percentage greater than 5
greater_5 = np.mean(exponential_data > 5) * 100

print(f"15. Percentage greater than 5: {greater_5:.2f}%")
```

```python
# ============================================
# C) COIN TOSS PROBABILITY EXPERIMENT
# ============================================

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

print("\n" + "=" * 60)
print("C) COIN TOSS PROBABILITY EXPERIMENT")
print("=" * 60)

# --------------------------------------------
# 1. Simulate 1,000 coin tosses
# --------------------------------------------

tosses = np.random.choice(
    ["Heads", "Tails"],
    size=1000
)

heads = np.sum(tosses == "Heads")
tails = np.sum(tosses == "Tails")

experimental_heads = heads / 1000
experimental_tails = tails / 1000

print("\n1. 1,000 Coin Tosses")
print(f"Total Tosses : 1000")
print(f"Heads        : {heads}")
print(f"Tails        : {tails}")
print(f"Experimental Probability of Heads : "
      f"{experimental_heads:.4f}")
print(f"Experimental Probability of Tails : "
      f"{experimental_tails:.4f}")

# 2. Theoretical probability
theoretical_heads = 0.5
theoretical_tails = 0.5

print("\n2. Theoretical Probability")
print(f"P(Heads) = {theoretical_heads:.4f}")
print(f"P(Tails) = {theoretical_tails:.4f}")

# 3. Bar chart
plt.figure(figsize=(7, 5))
plt.bar(
    ["Heads", "Tails"],
    [heads, tails]
)
plt.xlabel("Outcome")
plt.ylabel("Frequency")
plt.title("Frequency of Heads and Tails")
plt.show()

# 4. Pie chart
plt.figure(figsize=(7, 7))
plt.pie(
    [heads, tails],
    labels=["Heads", "Tails"],
    autopct="%1.1f%%"
)
plt.title("Percentage of Heads and Tails")
plt.show()

# 5. 10 tosses, 8 Heads
heads_10 = 8
total_10 = 10

experimental_probability_10 = heads_10 / total_10

print("\n5. Experimental Probability")
print("Total Tosses = 10")
print("Heads = 8")

print(
    f"Experimental Probability of Heads = "
    f"{heads_10}/{total_10} = "
    f"{experimental_probability_10:.2f}"
)

# --------------------------------------------
# 6. Simulate 10,000 coin tosses
# --------------------------------------------

tosses_10000 = np.random.choice(
    ["Heads", "Tails"],
    size=10000
)

heads_10000 = np.sum(tosses_10000 == "Heads")
tails_10000 = np.sum(tosses_10000 == "Tails")

exp_heads_10000 = heads_10000 / 10000
exp_tails_10000 = tails_10000 / 10000

diff_heads = abs(
    exp_heads_10000 - theoretical_heads
)

diff_tails = abs(
    exp_tails_10000 - theoretical_tails
)

print("\n" + "-" * 60)
print("6. COMPLETE ANALYSIS - 10,000 COIN TOSSES")
print("-" * 60)

print(f"Total Tosses                     : 10000")
print(f"Number of Heads                  : {heads_10000}")
print(f"Number of Tails                  : {tails_10000}")

print(f"\nExperimental Probability Heads: "
      f"{exp_heads_10000:.4f}")

print(f"Experimental Probability Tails: "
      f"{exp_tails_10000:.4f}")

print(f"\nTheoretical Probability Heads: "
      f"{theoretical_heads:.4f}")

print(f"Theoretical Probability Tails: "
      f"{theoretical_tails:.4f}")

print(f"\nDifference - Heads: "
      f"{diff_heads:.4f}")

print(f"Difference - Tails: "
      f"{diff_tails:.4f}")

# Bar chart for 10,000 tosses
plt.figure(figsize=(7, 5))

plt.bar(
    ["Heads", "Tails"],
    [heads_10000, tails_10000]
)

plt.xlabel("Outcome")
plt.ylabel("Frequency")
plt.title("Frequency of Heads and Tails - 10,000 Tosses")
plt.show()

# Pie chart for 10,000 tosses
plt.figure(figsize=(7, 7))

plt.pie(
    [heads_10000, tails_10000],
    labels=["Heads", "Tails"],
    autopct="%1.2f%%"
)

plt.title("Percentage Distribution - 10,000 Coin Tosses")
plt.show()
```
